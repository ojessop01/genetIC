```python
start = time.time()

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import simpson
import camb
from multiprocessing import Pool, cpu_count

C_KMS = 299792.458  # km/s

# =========================================================
# σ_vbc integrand (CAMB convention)
# =========================================================
def vbc_rms_using_Pr_from_Pm(kh, T_vbc_out, T_m_out, Pm, h):
    k = kh * h  # 1/Mpc

    # Undo CAMB /k^2 convention
    delta_m = T_m_out * k**2
    v_bc    = T_vbc_out * k**2

    # Primordial curvature spectrum
    P_R = Pm / (np.abs(delta_m)**2)

    # Velocity variance
    integrand = (k**2) * (np.abs(v_bc)**2) * P_R
    sigma2 = simpson(integrand, x=k) / (2.0 * np.pi**2)

    return np.sqrt(sigma2) * C_KMS


# =========================================================
# Chunked CAMB runner to bypass ~256 redshift hard limit
# =========================================================
def compute_sigma_chunk(z_chunk, base_pars, h):
    pars = base_pars.copy()
    pars.set_accuracy(AccuracyBoost=3, lAccuracyBoost=3, lSampleBoost=3)
    pars.set_matter_power(redshifts=z_chunk, kmax=kmax)

    results = camb.get_results(pars)
    transfer = results.get_matter_transfer_data()
    Pk_interp = results.get_matter_power_interpolator(
        hubble_units=False,
        k_hunit=True,
        nonlinear=False
    )
    sigma_vals = []

    for iz, z in enumerate(z_chunk):
        kh = transfer.transfer_z("k/h", iz)
        T_vbc_out = transfer.transfer_z("v_baryon_cdm", iz)
        T_b_out   = transfer.transfer_z("delta_baryon", iz)
        T_c_out = transfer.transfer_z("delta_cdm", iz)
        T_m_out = T_b_out*fb + fcdm*T_c_out
        Pm = Pk_interp.P(z, kh)

        sigma_vals.append(
            vbc_rms_using_Pr_from_Pm(kh, T_vbc_out, T_m_out, Pm, h)
        )

    return np.array(sigma_vals)

# =========================================================
# CLASS σ_vbc calc
# =========================================================
def sigma_vbc_CLASS_at_z(z):
    """
    Compute σ_vbc at a given z from CLASS transfer functions
    using the exact same physical definition as CAMB block.
    """
    trans = cosmo.get_transfer(z=z)
    k_h = trans['k (h/Mpc)']          # h/Mpc
    k = k_h * h                       # 1/Mpc

    theta_b  = trans['t_b']
    theta_c  = trans['t_cdm']
    theta_bc = theta_b - theta_c

    delta_m = trans['d_m']

    # Linear matter power at this z
    Pm = np.array([cosmo.pk(kk, z=z) for kk in k])

    # Primordial curvature spectrum
    P_R1 = As * (k / k_pivot)**(ns - 1)
    P_R2 = Pm / delta_m**2
    print(P_R2)
    print()
    print(P_R1)
    print()
    print(P_R2-P_R1)
    P_R = P_R1

    # Velocity: v = theta/k
    v_bc = theta_bc / k

    integrand = k**2 * np.abs(v_bc)**2 * P_R
    sigma2 = simpson(integrand, x=k) / (2*np.pi**2)

    return np.sqrt(sigma2) * C_KMS

# =========================================================
# Redshifts
# =========================================================
z_list = np.logspace(np.log10(1e-2), np.log10(1200), 10)
z_list = np.sort(z_list)[::-1]  # high → low

# =========================================================
# Run CAMB in chunks
# =========================================================
# MAX_PER_RUN = 256
# sigma_vals = []

# for i in range(0, len(z_list), MAX_PER_RUN):
#     z_chunk = z_list[i:i+MAX_PER_RUN]
#     sigma_vals.append(compute_sigma_chunk(z_chunk, base_pars, h))

# sigma_vals = np.concatenate(sigma_vals)

# =========================================================
# Run CLASS in parallel
# =========================================================
with Pool(cpu_count()) as P:
    sigma_class = np.array(P.map(sigma_vbc_CLASS_at_z, z_list))

print(sigma_class)

print(f'Finished Running, Total Run Time is: {time.time() - start:.2f}s')
```

```python
# =============================
# Correct slope computation
# =============================
ln_sigma = np.log(sigma_vals)
ln_sigma_class = np.log(sigma_class)

ln_1pz = np.log(1.0 + z_list)

# Work in ascending z so ln(1+z) increases
ln_sigma_inc       = ln_sigma[::-1]
ln_sigma_class_inc = ln_sigma_class[::-1]
ln_1pz_inc         = ln_1pz[::-1]

# Correct derivative
slope_inc       = np.gradient(ln_sigma_inc, ln_1pz_inc)
slope_class_inc = np.gradient(ln_sigma_class_inc, ln_1pz_inc)

# Flip back to match your descending-z plotting
slope       = slope_inc[::-1]
slope_class = slope_class_inc[::-1]

print(f"sigma_vbc(z≈1) (CLASS) = {sigma_class[-1]:.3f} km/s")

print(f"sigma_vbc(z≈1) (CAMB) = {sigma_vals[-1]:.3f} km/s")
# Plot σ_vbc vs z
# =========================================================
plt.figure(figsize=(8,6))
plt.plot(z_list, sigma_vals, lw=2, label=r"$\sigma_{v_{bc}}$ (CAMB)")
plt.plot(z_list, sigma_class, "--", lw=2,
         label=r"$\sigma_{v_{bc}}$ (CLASS)")

plt.axvline(z_rec, color="red", linestyle="--",
            label=fr"Recombination $z_* \approx {z_rec:.1f}$")
#plt.plot(z_PL, sigma_PL, "k:", lw=2, label=r"$\propto 1/(1+z)$")

plt.xlabel("z")
plt.ylabel(r"$\sigma_{v_{bc}}$ [km/s]")
plt.title("RMS Baryon–CDM Relative Velocity vs Redshift")
#plt.xlim(z_list.max(), None)
plt.ylim(0, None)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()


# =========================================================
# Plot slope
# =========================================================
plt.figure(figsize=(8,6))
plt.semilogx(z_list, slope, lw=2, label="CAMB")
plt.semilogx(z_list, slope_class, "--", lw=2, label="CLASS")

plt.axhline(1.0, color="k", linestyle=":",
            label="Expected slope = 1")
plt.axvline(z_rec, color="red", linestyle="--",
            label=fr"$z_* \approx {z_rec:.1f}$")

plt.xlabel("z")
plt.ylabel(r"$d\ln\sigma / d\ln(1+z)$")
plt.title("Instantaneous Decay Slope of $v_{bc}$")
#plt.xlim(z_list.max(), None)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#!/usr/bin/env python
import time
start = time.time()

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import simpson
import camb
from classy import Class
from multiprocessing import Pool, cpu_count

C_KMS = 299792.458  # km/s

# =========================================================
# Helpers: consistent k handling and CAMB /k^2 undo
# =========================================================
def kh_to_k(kh, h):
    """Convert k in (h/Mpc) convention to physical k in (1/Mpc)."""
    return kh * h

def camb_undo_k2(T_out, k):
    """
    Undo CAMB get_matter_transfer_data convention where many transfer outputs
    are stored with an implicit 1/k^2 factor.
    """
    return T_out * (k**2)

# =========================================================
# α(z) from transfer functions (consistent CAMB implementation)
# =========================================================
def alpha_from_CAMB(kh, T_b_out, T_c_out, T_m_out, h):
    # kh is k in (h/Mpc) convention; integrate over physical k (1/Mpc)
    k = kh_to_k(kh, h)

    # Undo CAMB /k^2 convention for density contrasts (dimensionless)
    delta_b = camb_undo_k2(T_b_out, k)
    delta_c = camb_undo_k2(T_c_out, k)
    delta_m = camb_undo_k2(T_m_out, k)

    delta_bc = delta_b - delta_c

    # Ensure purely real integrands (should be real anyway)
    num = simpson(k**2 * np.real(delta_bc * delta_m), x=k)
    den = simpson(k**2 * np.real(delta_m * delta_m), x=k)

    return num / den

# =========================================================
# CLASS α(z) (matched definitions to CAMB)
# =========================================================
def alpha_CLASS_at_z(z):
    trans = cosmo.get_transfer(z=z)

    kh = trans['k (h/Mpc)']          # (h/Mpc) convention
    k  = kh_to_k(kh, h)              # physical (1/Mpc)

    delta_b = trans['d_b']           # dimensionless
    delta_c = trans['d_cdm']         # dimensionless

    # Match CAMB definition of delta_m = f_b*delta_b + f_cdm*delta_cdm
    delta_m = fb * delta_b + fcdm * delta_c

    delta_bc = delta_b - delta_c

    num = simpson(k**2 * (delta_bc * delta_m), x=k)
    den = simpson(k**2 * (delta_m * delta_m), x=k)

    return num / den



# =========================================================
# CAMB chunk runner
# =========================================================
def compute_sigma_alpha_chunk(z_chunk, base_pars, h):
    pars = base_pars.copy()
    pars.set_accuracy(AccuracyBoost=3, lAccuracyBoost=3, lSampleBoost=3)
    pars.set_matter_power(redshifts=z_chunk, kmax=kmax)

    results = camb.get_results(pars)
    transfer = results.get_matter_transfer_data()

    Pk_interp = results.get_matter_power_interpolator(
        hubble_units=False,
        k_hunit=True,
        nonlinear=False
    )

    sigma_vals = []
    alpha_vals = []

    for iz, z in enumerate(z_chunk):
        kh = transfer.transfer_z("k/h", iz)

        T_vbc_out = transfer.transfer_z("v_baryon_cdm", iz)
        T_b_out   = transfer.transfer_z("delta_baryon", iz)
        T_c_out   = transfer.transfer_z("delta_cdm", iz)

        T_m_out = fb*T_b_out + fcdm*T_c_out
        Pm = Pk_interp.P(z, kh)

        sigma_vals.append(
            vbc_rms_using_Pr_from_Pm(kh, T_vbc_out, T_m_out, Pm, h)
        )

        alpha_vals.append(
            alpha_from_CAMB(kh, T_b_out, T_c_out, T_m_out, h)
        )

    return np.array(sigma_vals), np.array(alpha_vals)


# =========================================================
# Redshifts (high → low)
# =========================================================
z_list = np.logspace(np.log10(1), np.log10(200), 50)
z_list = np.sort(z_list)[::-1]


# =========================================================
# Run CAMB
# =========================================================
sigma_vals = []
alpha_vals = []

MAX_PER_RUN = 256

for i in range(0, len(z_list), MAX_PER_RUN):
    z_chunk = z_list[i:i+MAX_PER_RUN]
    s, a = compute_sigma_alpha_chunk(z_chunk, base_pars, h)
    sigma_vals.append(s)
    alpha_vals.append(a)

sigma_vals = np.concatenate(sigma_vals)
alpha_vals = np.concatenate(alpha_vals)

with Pool(cpu_count()) as P:
    alpha_class = np.array(P.map(alpha_CLASS_at_z, z_list))


print(f"Finished Running. Total Run Time: {time.time()-start:.2f}s")
```

```python

plt.figure(figsize=(8,6))
plt.plot(z_list, alpha_vals, lw=2, label=r"$\alpha(z)$ CAMB")
plt.plot(z_list, alpha_class, "--", lw=2, label=r"$\alpha(z)$ CLASS")
plt.gca().invert_xaxis()
plt.xlabel("z")
plt.ylabel(r"$\alpha(z)$")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

print(alpha_vals[-1],alpha_class[-1])
```
