```python
import numpy as np
import matplotlib.pyplot as plt

# Fourier-space spherical top-hat window
# ----------------------------------------------
def W_tophat(kr):
    return (np.sin(kr) - kr*np.cos(kr)) / (kr**3)


def rms_from_Pk_direct(k, Pk):
    """
    Compute RMS of an isotropic field directly from P(k),
    integrating in log-k space as a discrete sum.

    Parameters
    ----------
    k : array_like
        Wavenumbers in 1/Mpc (strictly positive, monotonic).
    Pk : array_like
        Power spectrum in Mpc^3.

    Returns
    -------
    sigma : float
        RMS of the field.
    sigma2 : float
        Variance of the field.
    """

    k = np.asarray(k)
    Pk = np.asarray(Pk)

    if np.any(k <= 0):
        raise ValueError("k must be strictly positive.")
    if k.ndim != 1 or Pk.ndim != 1 or len(k) != len(Pk):
        raise ValueError("k and Pk must be 1D arrays of equal length.")

    ln_k = np.log(k)

    # finite difference in log-space
    dlnk = np.diff(ln_k)

    # midpoint rule on log bins: use interior bins
    k_mid = np.sqrt(k[1:] * k[:-1])      # geometric midpoint
    P_mid = np.sqrt(Pk[1:] * Pk[:-1])    # geometric averaging is sensible in log space

    sigma2 = np.sum(P_mid * k_mid**3 * dlnk) / (2.0 * np.pi**2)

    # convert to km/s if this is a velocity spectrum
    sigma2 *= 299792.458**2

    return np.sqrt(sigma2), sigma2

# ----------------------------------------------
# σ8 normalisation factor at z = 0
# ----------------------------------------------
def sigma8_amplitude_normalisation_z0(
        k_hMpc,
        T_matter,
        ns,
        sigma8_target):
    """
    k_hMpc      : array of k in h/Mpc (from CAMB file)
    T_matter    : total-matter transfer function T_m(k)
    ns          : scalar spectral index
    sigma8_target : desired σ8
    """

    # σ8 from the raw transfer functions (no renormalisation)
    sigma8_pre = sigma_radius_from_transfers(
        radius_Mpc_h=8.0,
        k_array_hMpc=k_hMpc,
        T_array=T_matter,
        ns=ns
    )

    # Linear renormalisation
    linearRenorm = sigma8_target / sigma8_pre

    # Power spectrum amplitude rescales as δ²
    amplitude = linearRenorm**2

    return amplitude, sigma8_pre

z = 0
# ================================
# Transfers at z = 0 (undo /k^2)
# ================================
transfer = results.get_matter_transfer_data()

kh = transfer.transfer_z("k/h", z)
k  = kh * h

T_vbc = transfer.transfer_z("v_baryon_cdm", z) * k**2
T_m   = transfer.transfer_z("delta_tot", z) * k**2

# --- restrict to k < 100 ---
mask = k < 800
k     = k[mask]
kh    = kh[mask]
T_vbc = T_vbc[mask]
T_m   = T_m[mask]

# ================================
# Linear matter spectrum
# ================================
Pk_m = results.get_matter_power_interpolator(
    nonlinear=False,
    hubble_units=False,
    extrap_kmax = True, log_interp = True,
    k_hunit=False
).P(z, k)

# ================================
# Primordial curvature spectra
# ================================
P_R1 = As * (k / k_pivot)**(ns - 1.0) * ((2.0*np.pi**2) / k**3)
P_R2 = Pk_m / (np.abs(T_m)**2)

# ================================
# Method A and B for P_vbc
# ================================
P_vbc_1 = P_R1 * (np.abs(T_vbc)**2)
P_vbc_2 = P_R2 * (np.abs(T_vbc)**2)

# ================================
# Method C: direct CAMB P(k)
# ================================
PKint_vbc = results.get_matter_power_interpolator(
    nonlinear=False,
    hubble_units=False,
    k_hunit=False,extrap_kmax = True, log_interp = True,
    var1=13,
    var2=13
)

P_vbc_3 = PKint_vbc.P(z, k)

rms = rms_from_Pk_direct(k, P_vbc_3)
print(rms)

# ================================
# Plot comparison
# ================================
plt.figure(figsize=(7,6))
plt.loglog(k, P_vbc_1, label="Method A: As × Tvbc^2")
plt.loglog(k, P_vbc_2, '--', label="Method B: (Pm/Tm^2) × Tvbc^2")
plt.loglog(k, P_vbc_3, ':', label="Method C: CAMB direct")
plt.xlabel(r"$k\,[{\rm Mpc^{-1}}]$", fontsize = 20)
plt.ylabel(r"$P_{vbc}(k)$", fontsize = 20)
plt.title("vb–vc Power Spectrum Comparison (z = 0)", fontsize = 20)
plt.xlim(min(k),max(k))
plt.legend()
plt.show()

# ================================
# Absolute differences
# ================================
plt.figure(figsize=(7,6))
plt.loglog(k, np.abs(P_vbc_1 - P_vbc_3), label="|A − C|")
plt.loglog(k, np.abs(P_vbc_2 - P_vbc_3), label="|B − C|")
plt.xlabel(r"$k\,[{\rm Mpc^{-1}}]$", fontsize = 20)
plt.ylabel("Absolute Difference", fontsize = 20)
plt.title("Absolute Differences vs CAMB (z = 0)", fontsize = 20)
plt.xlim(min(k),max(k))
plt.legend()
plt.show()

# ================================
# Fractional differences
# ================================
plt.figure(figsize=(7,6))
plt.semilogx(k, P_vbc_1 / P_vbc_3 - 1.0, label="A / C − 1")
plt.semilogx(k, P_vbc_2 / P_vbc_3 - 1.0, label="B / C − 1")
plt.xlabel(r"$k\,[{\rm Mpc^{-1}}]$", fontsize = 20)
plt.ylabel("Fractional Difference", fontsize = 20)
plt.title("Fractional Differences vs CAMB (z = 0)", fontsize = 20)
plt.xlim(min(k),max(k))
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
from scipy.integrate import simpson
import camb

C_KMS = 299792.458
#print(z_list)

kmax = 800
# Use a fixed k grid in 1/Mpc

sigma_vals = []

transfer = results.get_matter_transfer_data()
for iz, z in enumerate(z_list):
    kh = transfer.transfer_z("k/h", iz)
    k  = kh * h

    # ---- mask k < 100 ----
    mask = k <= 1.0
    k = k[mask]

    T_vbc = transfer.transfer_z("v_baryon_cdm", iz)[mask] * k**2
    P_vbc = As * (k / k_pivot)**(ns - 1) * ((2.0*np.pi**2) / k**3) * T_vbc**2

    sigma, _ = rms_from_Pk_direct(k, P_vbc)
    sigma_vals.append(sigma)

sigma_vals = np.array(sigma_vals)

# for z, s in zip(z_list, sigma_vals):
#     print(f"σ_vbc(z={z:.0f}) = {s:.4f} km/s")

plt.figure(figsize=(7,6))
plt.plot(z_list, sigma_vals)
plt.xlabel("z")
plt.ylabel(r"$\sigma_{vbc}$  [km/s]")
plt.title("RMS baryon–CDM relative velocity vs redshift")

#plt.gca().invert_xaxis()   # <-- reverse x-axis direction
plt.xlim(max(z_list), 0)
plt.ylim(0, None)
plt.show()

ln_sigma = np.log10(sigma_vals)

ln_1pz = np.log10(1.0 + z_list)

# Work in ascending z so ln(1+z) increases
ln_sigma_inc       = ln_sigma[::-1]
ln_1pz_inc         = ln_1pz[::-1]

# Correct derivative
slope_inc       = np.gradient(ln_sigma_inc, ln_1pz_inc)

# Flip back to match your descending-z plotting
slope       = slope_inc[::-1]

# =========================================================
# Plot slope
# =========================================================
plt.figure(figsize=(8,6))
plt.plot(z_list, slope, lw=2)

plt.axhline(1.0, color="k", linestyle=":",
            label="Expected slope = 1")
#plt.axvline(z_rec, color="red", linestyle="--", label=fr"$z_* \approx {z_rec:.1f}$")

plt.xlabel("z")
plt.ylabel(r"$d\ln\sigma / d\ln(1+z)$")
plt.title("Instantaneous Decay Slope of $v_{bc}$")
#plt.xlim(z_list.max(), None)
plt.xlim(max(z_list), 0)
plt.legend()
plt.tight_layout()
plt.show()

# Mask to z < 200
mask = z_list <= 200

z_plot = z_list[mask]
slope_plot = slope[mask]

plt.figure(figsize=(8,6))
plt.plot(z_plot, slope_plot, lw=2)

plt.axhline(1.0, color="k", linestyle=":",
            label="Expected slope = 1")

plt.xlabel("z")
plt.ylabel(r"$d\ln\sigma / d\ln(1+z)$")
plt.title("Instantaneous Decay Slope of $v_{bc}$")

# Keep descending z-axis as before
plt.xlim(max(z_plot), 0)

plt.legend()
plt.tight_layout()
plt.show()

```
