```python
import matplotlib.pyplot as plt
from scipy.integrate import simpson
import camb

alpha_vals = []

def alpha_from_pk(k, P_bc_m, P_mm):
    """
    Compute α(z) using power spectra and discrete log-k integration:
    
        α = <δ_bc δ_m> / <δ_m δ_m>
          = ∫ k^3 P_bc,m dlnk / ∫ k^3 P_mm dlnk

    Parameters
    ----------
    k : array
        Wavenumbers in 1/Mpc (strictly positive, sorted).
    P_bc_m : array
        Cross power spectrum P_{δ_bc, δ_m}(k) [Mpc^3].
    P_mm : array
        Matter auto power spectrum P_{mm}(k) [Mpc^3].

    Returns
    -------
    alpha : float
    """

    k = np.asarray(k)
    P_bc_m = np.asarray(P_bc_m)
    P_mm = np.asarray(P_mm)

    if np.any(k <= 0):
        raise ValueError("k must be strictly positive.")

    ln_k = np.log(k)
    dlnk = np.diff(ln_k)

    # k^3 weighting (since ∫k²dk = ∫k³ dlnk)
    I_num = 0.5 * np.sum((k[1:]**3 * P_bc_m[1:] + k[:-1]**3 * P_bc_m[:-1]) * dlnk)
    I_den = 0.5 * np.sum((k[1:]**3 * P_mm[1:]   + k[:-1]**3 * P_mm[:-1])   * dlnk)

    return I_num / I_den



transfer = results.get_matter_transfer_data()
for iz, z in enumerate(z_list):
    kh = transfer.transfer_z("k/h", iz)
    k  = kh

    # ---- mask k < 100 ----
    mask = k <= 1
    k = k[mask]

    T_m = transfer.transfer_z("delta_nonu", iz)[mask] * k**2
    T_b = transfer.transfer_z("delta_baryon", iz)[mask] * k**2
    T_c = transfer.transfer_z("delta_cdm", iz)[mask] * k**2
    P_R = As * (k / k_pivot) ** (ns - 1) * ((2.0 * np.pi ** 2) / k ** 3)
    P_mm = P_R * T_m**2
    P_bcm = P_R * (T_b - T_c) * T_m

    alpha = alpha_from_pk(k, P_bcm, P_mm)
    alpha_vals.append(alpha)

alpha_vals = np.array(alpha_vals)

print(alpha_vals[0],alpha_vals[-1])

# for z, s in zip(z_list, sigma_vals):
#     print(f"σ_vbc(z={z:.0f}) = {s:.4f} km/s")

plt.figure(figsize=(7,6))
plt.plot(z_list, alpha_vals)
plt.xlabel("z")
plt.ylabel(r"$\alpha$")
plt.title(r"$\alpha$ vs redshift")

#plt.gca().invert_xaxis()   # <-- reverse x-axis direction
plt.xlim(max(z_list), 0)
#plt.ylim(0, None)
plt.show()


mask = z_list <= 200
plt.figure(figsize=(7,6))
plt.plot(z_list[mask], alpha_vals[mask])
plt.xlabel("z")
plt.ylabel(r"$\alpha$")
plt.title(r"$\alpha$ vs redshift")

#plt.gca().invert_xaxis()   # <-- reverse x-axis direction
plt.xlim(max(z_list[mask]), 0)
#plt.ylim(0, None)
plt.show()

ln_sigma = np.log10(np.abs(alpha_vals))

ln_1pz = np.log10(1.0 + z_list)

# Work in ascending z so ln(1+z) increases
ln_sigma_inc       = ln_sigma[::-1]
ln_1pz_inc         = ln_1pz[::-1]

# Correct derivative
slope_inc       = np.gradient(ln_sigma_inc, ln_1pz_inc)

# Flip back to match your descending-z plotting
slope       = slope_inc[::-1]

# =========================================================
# Plot slope
# =========================================================
plt.figure(figsize=(8,6))
plt.plot(z_list, slope, lw=2)

#plt.axhline(1.0, color="k", linestyle=":", label="Expected slope = 1")
#plt.axvline(z_rec, color="red", linestyle="--", label=fr"$z_* \approx {z_rec:.1f}$")

plt.xlabel("z")
plt.ylabel(r"$d\ln\alpha / d\ln(1+z)$")
plt.title(r"Instantaneous Decay Slope of $\alpha$")
#plt.xlim(z_list.max(), None)
plt.xlim(max(z_list), 0)
plt.legend()
plt.tight_layout()
plt.show()

# Mask to z < 200
mask = z_list <= 200

z_plot = z_list[mask]
slope_plot = slope[mask]

plt.figure(figsize=(8,6))
plt.plot(z_plot, slope_plot, lw=2)

#plt.axhline(1.0, color="k", linestyle=":",label="Expected slope = 1")

plt.xlabel("z")
plt.ylabel(r"$d\ln\alpha / d\ln(1+z)$")
plt.title(r"Instantaneous Decay Slope of $\alpha$")

# Keep descending z-axis as before
plt.xlim(max(z_plot), 0)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import camb

# ------------------------------------------------------------
# Alpha estimator
# ------------------------------------------------------------
def alpha_from_pk(k, P_bc_m, P_mm):
    k = np.asarray(k)
    P_bc_m = np.asarray(P_bc_m)
    P_mm = np.asarray(P_mm)

    if np.any(k <= 0):
        raise ValueError("k must be strictly positive")

    ln_k = np.log(k)
    dlnk = np.diff(ln_k)

    I_num = 0.5 * np.sum(
        (k[1:]**3 * P_bc_m[1:] + k[:-1]**3 * P_bc_m[:-1]) * dlnk
    )
    I_den = 0.5 * np.sum(
        (k[1:]**3 * P_mm[1:] + k[:-1]**3 * P_mm[:-1]) * dlnk
    )

    return I_num / I_den


# ------------------------------------------------------------
# CAMB transfer data (assumed already computed)
# ------------------------------------------------------------
# results : camb.CAMBresults
# As, ns, k_pivot defined consistently with CAMB

transfer = results.get_matter_transfer_data()
#print(z_list[-1])
#print(z_list)
# ------------------------------------------------------------
# Select z = 0
# ------------------------------------------------------------
z_target = 0.0
iz = 255

kh = transfer.transfer_z("k/h", iz)
k  = kh

T_m = transfer.transfer_z("delta_nonu", iz) * k**2
T_b = transfer.transfer_z("delta_baryon", iz) * k**2
T_c = transfer.transfer_z("delta_cdm", iz) * k**2

P_R = As * (k / k_pivot)**(ns - 1) * (2.0 * np.pi**2 / k**3)

P_mm  = P_R * T_m**2
P_bcm = P_R * (T_b - T_c) * T_m


# ------------------------------------------------------------
# k_max scan
# ------------------------------------------------------------
kmax_list = np.logspace(-4, 2.5, 500)  # 0.1 → 10 Mpc⁻¹
alpha_kmax = np.zeros_like(kmax_list)

for i, kmax in enumerate(kmax_list):
    mask = k <= kmax
    alpha_kmax[i] = alpha_from_pk(
        k[mask],
        P_bcm[mask],
        P_mm[mask]
    )


# ------------------------------------------------------------
# Convergence plot
# ------------------------------------------------------------
plt.figure(figsize=(7, 5))
plt.plot(kmax_list, alpha_kmax, lw=2)
plt.xscale("log")
plt.xlabel(r"$k_{\max}\;[\mathrm{Mpc}^{-1}]$")
plt.ylabel(r"$\alpha(z=0)$")
plt.title(r"Convergence of $\alpha$ with UV cutoff at $z=0$")
plt.xlim(kmax_list.min(),kmax_list.max())
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Optional: fractional convergence relative to largest k_max
# ------------------------------------------------------------
alpha_inf = alpha_kmax[-1]

plt.figure(figsize=(7, 5))
plt.plot(
    kmax_list,
    (alpha_kmax - alpha_inf) / alpha_inf,
    lw=2
)
plt.xscale("log")
plt.xlabel(r"$k_{\max}\;[\mathrm{Mpc}^{-1}]$")
plt.ylabel(r"$(\alpha - \alpha_{\infty}) / \alpha_{\infty}$")
plt.title(r"UV sensitivity of $\alpha$ at $z=0$")
plt.axhline(0, color="k", ls="--", lw=1)
plt.xlim(kmax_list.min(),kmax_list.max())
plt.tight_layout()
plt.show()
```

```python
# ------------------------------------------------------------
# k_min scan (IR cutoff)
# ------------------------------------------------------------
kmax_fixed = 100.0  # Mpc^-1
kmin_list  = np.logspace(-4, 1, 400)  # 1e-4 → 0.1 Mpc^-1

alpha_kmin = np.zeros_like(kmin_list)

for i, kmin in enumerate(kmin_list):
    mask = (k >= kmin) & (k <= kmax_fixed)
    alpha_kmin[i] = alpha_from_pk(
        k[mask],
        P_bcm[mask],
        P_mm[mask]
    )


# ------------------------------------------------------------
# Convergence plot (IR cutoff)
# ------------------------------------------------------------
plt.figure(figsize=(7, 5))
plt.plot(kmin_list, alpha_kmin, lw=2)
plt.xscale("log")
plt.xlabel(r"$k_{\min}\;[\mathrm{Mpc}^{-1}]$")
plt.ylabel(r"$\alpha(z=0)$")
plt.title(r"Convergence of $\alpha$ with IR cutoff ($k_{\max}=10$)")
plt.xlim(kmin_list.min(), kmin_list.max())
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Fractional deviation relative to smallest k_min
# ------------------------------------------------------------
alpha_inf = alpha_kmin[0]

plt.figure(figsize=(7, 5))
plt.plot(
    kmin_list,
    (alpha_kmin - alpha_inf) / alpha_inf,
    lw=2
)
plt.xscale("log")
plt.xlabel(r"$k_{\min}\;[\mathrm{Mpc}^{-1}]$")
plt.ylabel(r"$(\alpha - \alpha_{\infty}) / \alpha_{\infty}$")
plt.title(r"IR sensitivity of $\alpha$ at $z=0$ ($k_{\max}=10$)")
plt.axhline(0, color="k", ls="--", lw=1)
plt.xlim(kmin_list.min(), kmin_list.max())
plt.tight_layout()
plt.show()
```
